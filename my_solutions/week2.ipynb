```python
sample_space = itertools.product(range(1,7),repeat=2)
n_tot = 0
n_7 = 0

for outcome in sample_space:
    n_tot += 1
    if sum(outcome) == 7:
        n_7 += 1

print(f'Pr(sum = 7) = {n_7/n_tot}')
```

```python
sample_space = itertools.product(range(1,7),repeat=2)
n_tot = 0
n_A = 0
n_B = 0
n_AB = 0
for outcome in sample_space:
    n_tot += 1
    if sum(outcome) == 7:
        n_A += 1
    if outcome[0] == 4 or outcome[1] == 4:
        n_B += 1
        if sum(outcome) == 7:
            n_AB += 1

print(f'Pr(A|B) = {n_AB/n_tot}')
print(f'Pr(A)   = {n_A/n_tot}')
```
