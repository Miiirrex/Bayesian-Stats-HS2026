## Pi estimation

### Numpy's Pi definition

```python
numpy_pi = np.pi
print(f"Numpy Pi: {numpy_pi}")
```

### Analytic Series Expansion

```python
N = 10000
ana_pi = np.sum([4 * (-1)**k / (2*k + 1) for k in range(N)])
print(f"Analytic Series Expansion of Pi with N = {N}: {ana_pi}")
```

```python
ana_pi_plot = np.zeros(N)
for i in range(N):
    ana_pi_plot[i] = np.sum([4 * (-1)**k / (2*k + 1) for k in range(i)])

fig,ax = plt.subplots(figsize=(8,6))

ax.plot(range(N),ana_pi_plot-np.pi,label='Analytic Series Expansion difference to Pi',color='red')
ax.axhline(0,label='Numpy Pi',color='grey',ls='--')
ax.set_xlim((1,50))
ax.set_ylim((-1,1))
ax.set_title("Analytic Series Expansion Approach to Pi")
ax.set_xlabel("Number of Summands")
ax.set_ylabel("Difference to Pi")
ax.grid()
ax.legend()
```

### Monte-Carlo Estimate of Pi

```python
n = 10000
mc_unitsq = np.random.uniform(-1.0,1.0,(n,2))
nr = 0
mc_unitdisc = np.zeros([1,2])

for i in range(n):
    if np.sqrt(mc_unitsq[i,0]**2 + mc_unitsq[i,1]**2) <= 1.0:
        mc_unitdisc = np.append(mc_unitdisc,[[mc_unitsq[i,0],mc_unitsq[i,1]]])
        nr += 1

print(4 * (nr/n))
print(mc_unitdisc)
```

```python
fig,ax = plt.subplots(figsize=(8,8))

ax.plot(mc_unitsq[:,0],mc_unitsq[:,1],marker='.',ls='')
ax.plot(mc_unitdisc[::2],mc_unitdisc[1::2],marker='.',ls='')
```
