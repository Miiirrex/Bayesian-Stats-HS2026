### Exercise 1

Consider the distribution with PDF
$$
    p(x) = \begin{cases}
                \frac{1}{2}\sin x & 0 \le x \le \pi \\
                0 & \text{otherwise}
            \end{cases}
$$

1. Derive the CDF $F(x) = \int_{-\infty}^{x} p(x')\,\dd x'$ analytically, and plot it together with the PDF $p(x)$.
2. Find the inverse CDF $F^{-1}(u)$, and use it to transform samples $U \sim \Uniform(0, 1)$ into samples $X = F^{-1}(U)$ (inverse transform sampling).
3. Compare a density-normalised histogram of the samples with the PDF $p(x)$.

```python
def p1(x):
    x = np.asarray(x)
    return np.where((x >= 0) & (x <= np.pi), 0.5 * np.sin(x), 0.0)

def F1(x):
    x = np.asarray(x)
    return np.where(x < 0, 0.0,
           np.where(x > np.pi, 1.0, 0.5 * (1 - np.cos(x))))
```

```python
x = np.linspace(-0.5,np.pi+0.5,100)

plt.title('PDF vs CDF')
plt.plot(x,p1(x),label='PDF')
plt.plot(x,F1(x),label='CDF')
plt.grid()
plt.legend()
```

```python
def F_inv(y):
    y = np.asarray(y)
    return  np.where(y < 0.0, 0,
            np.where((y >= 0) & (y <= 1), np.arccos(1 - 2 * y), np.pi))
```

```python
u = np.linspace(0.01,0.99,80)
X = F_inv(u)

fig, ax= plt.subplots()

ax.plot(x,p1(x),label='PDF', ls='--',color='orange')
ax.plot(x,F1(x),label='CDF')
ax.hist(X, density=True, alpha=1.0, label="Histogram of $Y$", histtype="step", color="C1")
ax.hlines(y=u, xmin=-1, xmax=X, lw=1, alpha=0.5, color="grey")
ax.vlines(x=X, ymin=0, ymax=u, lw=1, alpha=0.5, color="grey", label="$y = F_Y^{-1}(u)$")

ax.legend(frameon=False, loc=2, bbox_to_anchor=(1.05, 1.))
```
